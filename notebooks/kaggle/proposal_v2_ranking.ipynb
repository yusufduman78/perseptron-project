# Perseptron v2 - MAP@12 Ranking Evaluation

Runs proposal-aligned MAP@12, Precision@10, Recall@10, HitRate@12, and CandidateRecall evaluation for a selected customer-level fold.

```python
from pathlib import Path
import os
import subprocess
import sys

candidates = [Path.cwd(), Path('/kaggle/working/perseptron_project'), Path('/kaggle/input/perseptron-project')]
PROJECT_DIR = next((path for path in candidates if (path / 'src' / 'proposal_v2').exists()), Path.cwd())
os.environ['PYTHONPATH'] = str(PROJECT_DIR)
print('PROJECT_DIR =', PROJECT_DIR)

def run_module(module, *args):
    command = [sys.executable, '-m', module, *map(str, args)]
    print('RUN:', ' '.join(command))
    subprocess.run(command, cwd=PROJECT_DIR, check=True)
```

```python
FAST_RUN = True
FOLD_ID = 0
SAMPLE_CUSTOMERS = 100 if FAST_RUN else None

args = ['--fold-id', FOLD_ID, '--candidate-limit', 5000, '--visual-neighbors', 3000, '--co-purchase-per-item', 300]
if SAMPLE_CUSTOMERS is not None:
    args += ['--sample-customers', SAMPLE_CUSTOMERS]
run_module('src.proposal_v2.ranking', *args)
run_module('src.proposal_v2.aggregate_results')
```
