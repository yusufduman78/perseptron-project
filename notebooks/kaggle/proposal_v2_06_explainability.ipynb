# Perseptron v2 - 06 Explainability: SHAP / Permutation + Grad-CAM

Bu notebook proposal'daki explainability beklentisini kapatir.

- Tabular taraf: SHAP calisabiliyorsa SHAP summary, calismazsa ayni checkpoint uzerinde permutation fallback.
- Gorsel taraf: image-only EfficientNet-B0 CNN baseline uzerinden Grad-CAM ornekleri.
- Late-fusion icin yorum: Grad-CAM CNN baseline'i aciklar; late-fusion visual branch icin embedding-space similarity ayrica raporlanir.

```python
from pathlib import Path
import os, subprocess, sys
CANDIDATES = [Path.cwd(), Path('/kaggle/working/perseptron_project'), Path('/kaggle/input/perseptron-project')]
PROJECT_DIR = next((path for path in CANDIDATES if (path / 'src' / 'proposal_v2').exists()), Path.cwd())
os.environ['PYTHONPATH'] = str(PROJECT_DIR)
print('PROJECT_DIR =', PROJECT_DIR)
def run_module(module, *args):
    command = [sys.executable, '-m', module, *map(str, args)]
    print('RUN:', ' '.join(command))
    subprocess.run(command, cwd=PROJECT_DIR, check=True)
```

## Parametreler

Bu notebook en az `tabular_only_fold0.pt` ve `image_only_effnet_cnn_fold0.pt` olustuktan sonra calistirilir.

```python
FOLD_ID = 0
SHAP_BACKGROUND_ROWS = 80
SHAP_EXPLAIN_ROWS = 40
SHAP_NSAMPLES = 100
MAX_GRADCAM_EXAMPLES = 12
```

## Explainability Ciktilari

Ciktilar `reports/proposal_v2/` altina yazilir.

```python
run_module(
    'src.proposal_v2.explainability',
    '--fold-id', FOLD_ID,
    '--shap-background-rows', SHAP_BACKGROUND_ROWS,
    '--shap-explain-rows', SHAP_EXPLAIN_ROWS,
    '--shap-nsamples', SHAP_NSAMPLES,
    '--max-gradcam-examples', MAX_GRADCAM_EXAMPLES,
)
```
