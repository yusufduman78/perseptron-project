# Perseptron v2 - Proposal Aligned Training

This notebook runs the proposal-aligned v2 experiments on Kaggle: customer-level fold creation, tabular-only, image-history, multimodal late-fusion, and optional EfficientNet-B0 image-only CNN baseline.

```python
from pathlib import Path
import os
import subprocess
import sys

candidates = [Path.cwd(), Path('/kaggle/working/perseptron_project'), Path('/kaggle/input/perseptron-project')]
PROJECT_DIR = next((path for path in candidates if (path / 'src' / 'proposal_v2').exists()), Path.cwd())
os.environ['PYTHONPATH'] = str(PROJECT_DIR)
print('PROJECT_DIR =', PROJECT_DIR)

def run_module(module, *args):
    command = [sys.executable, '-m', module, *map(str, args)]
    print('RUN:', ' '.join(command))
    subprocess.run(command, cwd=PROJECT_DIR, check=True)
```

```python
# Parameters
FAST_RUN = True
FOLD_ID = 0
RUN_SPLITS = True
RUN_MLP_MODELS = True
RUN_CNN_BASELINE = True

MAX_TRAIN_POSITIVES = 5_000 if FAST_RUN else 200_000
MAX_VAL_POSITIVES = 1_000 if FAST_RUN else 50_000
EPOCHS = 1 if FAST_RUN else 3
CNN_MAX_TRAIN_POSITIVES = 1_000 if FAST_RUN else 20_000
CNN_MAX_VAL_POSITIVES = 300 if FAST_RUN else 5_000
CNN_EPOCHS = 1 if FAST_RUN else 2
```

```python
if RUN_SPLITS:
    run_module('src.proposal_v2.splits')

if RUN_MLP_MODELS:
    run_module(
        'src.proposal_v2.train',
        '--fold-id', FOLD_ID,
        '--models', 'tabular_only,image_history,late_fusion',
        '--max-train-positives', MAX_TRAIN_POSITIVES,
        '--max-val-positives', MAX_VAL_POSITIVES,
        '--epochs', EPOCHS,
    )

if RUN_CNN_BASELINE:
    run_module(
        'src.proposal_v2.cnn_baseline',
        '--fold-id', FOLD_ID,
        '--max-train-positives', CNN_MAX_TRAIN_POSITIVES,
        '--max-val-positives', CNN_MAX_VAL_POSITIVES,
        '--epochs', CNN_EPOCHS,
    )
```
